```python
                # INTERNSHIP APPLICATION TRACKER
applications = []

# APPLICATION ADDITION
def add_application():
    company = input("Company name: ")
    role = input("Role: ")
    stipend = input("Stipend: ")
    deadline = input("Deadline: ")

    application = {
        "company": company,
        "role": role,
        "stipend": stipend,
        "deadline": deadline,
        "status": "Applied"
    }

    applications.append(application)
    print("Application added successfully!")
    
# VIEWING APPLICATIONS 
def view_applications():
    if len(applications) == 0:
        print("No applications added yet.")
        return

    for index, application in enumerate(applications, start=1):
        print("\nApplication", index)
        print("Company:", application["company"])
        print("Role:", application["role"])
        print("Stipend:", application["stipend"])
        print("Deadline:", application["deadline"])
        print("Status:", application["status"])

while True:
    print("\n--- Internship Application Tracker ---")
    print("1. Add application")
    print("2. View all applications")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_application()
    elif choice == "2":
        view_applications()
    elif choice == "3":
        print("Good luck with your applications!")
        break
    else:
        print("Invalid choice. Please enter 1, 2, or 3.")
```
